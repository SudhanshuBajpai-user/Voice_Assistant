```python
import whisper
import torch
from pathlib import Path

MODEL_NAME = "base.en"

AUDIO_FILE = Path(
    "/home/sudhanshu_bajpai/projects/Voice_assistant/audio/command.wav"
)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", DEVICE)
print("Loading Whisper...")

model = whisper.load_model(
    MODEL_NAME,
    device=DEVICE
)

print("Whisper loaded.")
print("Transcribing...")

result = model.transcribe(
    str(AUDIO_FILE),
    fp16=(DEVICE == "cuda")
)

text = result["text"].strip()

print(f"You said: {text}")
```
